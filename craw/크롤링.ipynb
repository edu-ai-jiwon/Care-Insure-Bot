```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://tricare.mil/Costs/Compare"

def quick_crawl():
    res = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
    soup = BeautifulSoup(res.text, 'html.parser')
    
    rows = soup.select("#content table tbody tr")
    
    return [
        {
            "비용 유형": r.select_one("td:nth-child(1)").get_text(strip=True),
            "TRICARE Prime - A그룹": r.select_one("td:nth-child(2)").get_text(" ", strip=True),
            "TRICARE Select - B그룹": r.select_one("td:nth-child(3)").get_text(" ", strip=True)
        }
        for r in rows if r.select_one("td")
    ]


# 1. 함수를 실행하고 그 결과값(리스트)을 'data' 변수에 할당
data = quick_crawl()

# 2. 이제 'data' 변수가 존재하므로 판다스에서 사용할 수 있음
df = pd.DataFrame(data)

# 3. CSV 파일로 저장
df.to_csv("Health_Plan_Costs.csv", index=False, encoding='utf-8-sig')

print("파일 저장 완료: Health_Plan_Costs.csv")
```

```python
import os
import pandas as pd
from dotenv import load_dotenv
# 이 부분을 아래와 같이 수정해 보세요.
from langchain_core.documents import Document 

# 1. .env 파일에서 OPENAI_API_KEY 로드
load_dotenv()

# 2. CSV 파일 읽기
try:
    df = pd.read_csv("data/mental_health_services.csv")

    # 3. 데이터프레임의 각 행을 Document 객체로 변환
    documents = []
    for _, row in df.iterrows():
        content = f"비용 유형: {row['서비용 유형']}\n상세 내용: {row['내용']}"
        documents.append(Document(page_content=content, metadata={"source": "tricare_csv"}))

    print(f"✅ 로드 완료: {len(documents)}개의 서비스 항목을 불러왔습니다.")
    
except Exception as e:
    print(f"❌ 오류 발생: {e}")
```

# 텍스트 분할 (Chunking)
- AI 모델은 한 번에 읽을 수 있는 양이 정해져 있고, 너무 긴 글은 검색 효율을 떨어짐
- 따라서 문서를 적절한 크기의 '덩어리(Chunk)'로 자르는 과정이 필요

```python
# 경로를 langchain_text_splitters로 변경했습니다.
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 문서를 의미 있는 단위로 쪼개는 설정
# chunk_size: 한 덩어리의 글자 수 (약 500자 내외)
# chunk_overlap: 앞뒤 덩어리가 겹치는 부분 (문맥의 연결성을 유지하기 위함)
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

# 이전 단계에서 만든 documents를 나눕니다.
splits = text_splitter.split_documents(documents)

print(f"✅ 분할 완료: 원본 문서가 {len(splits)}개의 청크(덩어리)로 나뉘었습니다.")
```

# 임베딩 및 벡터 데이터베이스 저장 (Embedding & Storage)
- 나뉘어진 텍스트들을 숫자로 변환(Embedding)하여 지도상의 좌표처럼 저장하는 단계
- 이를 통해 '키워드'가 아닌 '의미'로 검색이 가능

```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma

# OpenAI의 최신 임베딩 모델 사용
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")

# Chroma DB에 벡터화하여 저장 (로컬 폴더 'tricare_vdb'에 물리적 저장)
vectorstore = Chroma.from_documents(
    documents=splits, 
    embedding=embedding_model,
    persist_directory="./tricare_vdb"
)

print("벡터 DB 구축 완료: 모든 데이터가 벡터화되어 저장되었습니다.")
```

# 질문 검색 및 답변 생성 (Retrieval & Generation)
- 사용자의 질문을 받아서 DB에서 관련 정보를 찾아낸 뒤, LLM(GPT)이 그 정보를 바탕으로 답변을 구성하는 최종 단계

```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_community.vectorstores import Chroma

# 1. 벡터 DB 불러오기 (이전 단계에서 저장한 경로 사용)
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = Chroma(persist_directory="./tricare_vdb", embedding_function=embedding_model)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

# 2. LLM 설정
llm = ChatOpenAI(model_name="gpt-4o-mini", temperature=0)

# 3. 프롬프트 설정
template = """아래의 문맥만을 사용하여 질문에 답하세요:
{context}

질문: {question}

답변 (한국어로 정중하게):"""
prompt = ChatPromptTemplate.from_template(template)

# 4. RAG 파이프라인 구성 (LCEL 방식 - chains 모듈 불필요)
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

# 5. 질문 실행
query = "심리 검사(Psychological testing)가 보험 적용이 안 되는 경우는 언제야?"
try:
    response = rag_chain.invoke(query)
    print(f"\n[질문]: {query}")
    print(f"\n[AI 답변]:\n{response}")
except Exception as e:
    print(f"❌ 실행 중 오류 발생: {e}")
```

# multilingual-e5-large 모델의 특징

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. 사진 속 모델(multilingual-e5-large) 설정
# 로컬 메모리를 사용하므로 처음 실행 시 모델 다운로드 시간이 약간 소요됩니다.
model_name = "intfloat/multilingual-e5-large"
encode_kwargs = {'normalize_embeddings': True} # 성능 향상을 위한 정규화

embedding_model = HuggingFaceEmbeddings(
    model_name=model_name,
    encode_kwargs=encode_kwargs
)

# 2. 벡터 DB 구축 (기존 DB와 섞이지 않도록 경로를 새로 지정하는 것이 좋습니다)
# 임베딩 모델이 바뀌면 벡터의 차원이 달라지므로 새로 구축해야 합니다.
vectorstore = Chroma.from_documents(
    documents=splits, # 이전에 만든 splits 변수 사용
    embedding=embedding_model,
    persist_directory="./tricare_e5_vdb"
)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})

# 3. LLM 및 프롬프트 설정 (이전과 동일)
llm = ChatOpenAI(model_name="gpt-4o-mini", temperature=0)

template = """아래의 문맥(context)만을 사용하여 질문에 답하세요:
{context}

질문: {question}

답변 (한국어로 정중하게):"""
prompt = ChatPromptTemplate.from_template(template)

# 4. RAG 파이프라인 구성 (LCEL)
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

# 5. 실행
query = "심리 검사 보장이 안 되는 구체적인 예시들을 알려줘."
response = rag_chain.invoke(query)

print(f"\n[모델]: {model_name}")
print(f"[질문]: {query}")
print(f"[AI 답변]:\n{response}")
```
